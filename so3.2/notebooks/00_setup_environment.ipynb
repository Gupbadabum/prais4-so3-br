## Google Drive

O Google Drive é utilizado como repositório dos dados de entrada e dos
produtos geoespaciais que, devido ao volume, não são armazenados no
repositório GitHub.

```python
from google.colab import drive

drive.mount("/content/drive")

DRIVE_ROOT = Path("/content/drive/MyDrive/Cemaden")
PROJECT_ROOT = DRIVE_ROOT / "PRAIS4_SO3_BR"

SO31_ROOT = PROJECT_ROOT / "SO3.1"
SO32_ROOT = PROJECT_ROOT / "SO3.2"
SO33_ROOT = PROJECT_ROOT / "SO3.3"

SPEI_DIR = SO31_ROOT / "outputs" / "spei12_classified"

SO32_INPUT = SO32_ROOT / "input"
SO32_INTERMEDIATE = SO32_ROOT / "intermediate"
SO32_OUTPUT = SO32_ROOT / "outputs"
SO32_LOGS = SO32_ROOT / "logs"
```

```python
spei_files = sorted(SPEI_DIR.glob("*.tif"))

print(f"GeoTIFFs SPEI12 encontrados: {len(spei_files)}")
```

```python
directories = {
    "Projeto": PROJECT_ROOT,
    "SO3.1": SO31_ROOT,
    "SPEI12": SPEI_DIR,
    "SO3.2 input": SO32_INPUT,
    "SO3.2 intermediate": SO32_INTERMEDIATE,
    "SO3.2 output": SO32_OUTPUT,
    "SO3.2 logs": SO32_LOGS,
}

for name, path in directories.items():
    status = "OK" if path.exists() else "NÃO ENCONTRADO"
    print(f"{name:20s}: {status} | {path}")
```

## Google Earth Engine
